#  Lists

```python
fruit_selection = ["apple", "banana", "cherry", "date", "elderberry"]
```

```python
# 1.Meyvələri əlifba sırasına görə sıralamaq üçün bir Python kodu yazın. Sıralanmış siyahını çap edin.
# burda əvvlcə reverse etdim çünki list onsuzda əlifba sıralaması ilə idi

fruit_selection.reverse()
fruit_selection.sort()
print(fruit_selection)


```

```python
# 2. fruit_selection siyahısının sonuna "orange" meyvəsini əlavə edən bir Python kodu yazın. Daha sonra yenilənmiş siyahını çap edin.
fruit_selection.append("orange")
print(fruit_selection)
```

```python
# 2.1 fruit_selection siyahısının sonuna "heyva" meyvəsini əlavə edən bir Python kodu yazın. Daha sonra yenilənmiş siyahını çap edin.
# burda mən exteddən istifadə etdim əgər appenddə kimi yaılmış olsaydı bu zaman heyva sözü hərflər kimi daxil olardı
fruit_selection.extend(["heyva"])
print(fruit_selection)
```

```python
# 3. fruit_selection siyahısının son elementini silin və onu çap edin.

fruit_selection.remove('heyva')
print(fruit_selection)
```

```python
# 3.1 fruit_selection siyahısının son elementini silin və onu çap edin.

fruit_selection.remove(fruit_selection[5])
print(fruit_selection)
```

# Dictionaries

```python
product_dict = {
    "Laptop": ["Dell", 999.99],
    "Smartphone": ["Apple", 799.99],
    "Headphones": ["Sony", 129.95],
    "Tablet": ["Samsung", 349.99]
}
```

```python
# 1.product_dict lüğətinin açarlarını (keys) və dəyərlərini (values) çap edin.
```

```python
# 1.product_dict lüğətinin açarlarını (keys) və dəyərlərini (values) çap edin.

print(f"acarlar: {product_dict.keys()}")
print(f"deyerler: {product_dict.values()}")
```

# Tuple

```python
my_tuple = ("Auto", "Fahrrad", "Motorrad")
```

```python
# 1. Python-da my_tuple tuplenin ilk və son elementini çap edin.

print(my_tuple[0])
print(my_tuple[-1])
```

```python
# 2 Python-da my_tuple  tuplenin ilk elementini silin.
s=list(my_tuple)
s.remove("Auto")
print(s)
```

```python
# 2.1 Python-da my_tuple  tuplenin ilk elementini silin.
s1 = list(my_tuple)
s1.remove(s1[0])
print(s1)
```

```python
# 2.2 Python-da my_tuple  tuplenin ilk elementini silin.
s2=list(my_tuple)
s2.remove(s2[s2.index("Auto")])

print(s2)
```
