```python
# 9. Aşağıda verilmiş siyahıda (list) 2-ci elementi dəyişdirin və siyahını çap edin (print)
meyveler = ["alma", "banan", "gilas"]
```

```python
# append ancaq bize icaze verir ki 1 argument elave edek 1den cox argumenti
# ancaq [] bu isarenin icine yazdiqda elave etmek olur list icinde list kimi

meyveler.append("armud")
print(f"append: {meyveler}")
```

```python
#extend 1den cox argument elae etmeye icaze verir list icine

meyveler.extend(["heyva","yalaci alma"])
print(f"extend: {meyveler}")
```

```python
# insert ile hem elave etmek istediyimiz argumenti hem de hara elave etmek
# istiyirik onu yaza bilirik

meyveler.insert(0,"edalet")

print(f"insert: {meyveler}")
```

```python
# remove vasitesile list daxilinden istediyimiz argumentin adini yazaraq c;xarda bilirik

meyveler.remove("edalet")

print(f"remove: {meyveler}")
```

```python
# pop vasitesile argumenti indeksi vasitesile sile bilirik

meyveler.pop(1)

print(f"pop: {meyveler}")
```

```python
# index-e esasen deyeri yeni deyerle deyisirik

meyveler[1]="qarpiz"

print(f"yeni deyer: {meyveler}")
```

```python
# clear vasitesile bwtwn list hamsi silinir

meyveler.clear()

print(f"clear: {meyveler}")
```
